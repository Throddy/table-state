# Трек C: офлайн-веса Grounding DINO и OWLv2 (`ts-models-ov`)

Готовит каталог весов для ноутбуков с Internet `Off`. GPU не нужен, достаточно CPU-сессии.

**Перед запуском**

1. Settings: Accelerator `None`, Internet `On` (только этот ноутбук).
2. Add Input: датасет с кодом репозитория (коммит с `tracks/ov/pack_weights.py`). Данные соревнования не нужны.
3. Save Version → Save & Run All.

**Что делает** `python tracks/ov/pack_weights.py --out /kaggle/working/ts-models-ov`:

- скачивает обе модели с зафиксированными ревизиями из `tracks/ov/fetch_weights.py` (около 2.7 ГБ);
- сверяет sha256 `model.safetensors` с Hugging Face;
- в отдельном процессе без сети загружает процессор, токенизатор и модель кодом `run_ov.py`;
- пишет `MANIFEST.json` и `README.md`.

**После прогона**: вкладка Output этой версии → `New Dataset` из каталога `ts-models-ov`, имя `ts-models-ov`, Private. Дальше датасет подключается как Input в ноутбуках трека C, `kaggle_setup.py` находит веса сам.

```python
import os
from pathlib import Path

found = sorted(p.parents[2] for p in Path('/kaggle/input').rglob('tracks/ov/pack_weights.py'))
assert len(found) == 1, f'Нужен ровно один подключённый датасет с кодом, где есть tracks/ov/pack_weights.py, найдено: {found}'
os.chdir(found[0])
os.environ['TS_ARTIFACTS'] = '/kaggle/working/artifacts'
print('код:', found[0])
```
